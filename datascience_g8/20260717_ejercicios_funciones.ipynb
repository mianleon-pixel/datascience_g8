```python
#Ejercicio 1
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
def analizar_ventas(ventas, meta=None):
   
    if len(ventas) == 0:
        return "Error: La lista está vacía o contiene montos negativos."
    
   
    for v in ventas:
        if v < 0:
            return "Error: La lista está vacía o contiene montos negativos."
            
  
    total = 0
    maxima = ventas[0]
    minima = ventas[0]
    debajo_100 = 0    
   
    for v in ventas:
        total += v
        
        if v > maxima:
            maxima = v
            
        if v < minima:
            minima = v
            
        if v < 100:
            debajo_100 += 1            
    
    promedio = total / len(ventas)    
    
    superiores_promedio = 0
    for v in ventas:
        if v > promedio:
            superiores_promedio += 1            
 
    resumen = {
        "total": total,
        "promedio": promedio,
        "maxima": maxima,
        "minima": minima,
        "superiores_promedio": superiores_promedio,
        "debajo_100": debajo_100
    }    
  
    if meta is not None:
        if total >= meta:
            resumen["meta_alcanzada"] = True
        else:
            resumen["meta_alcanzada"] = False
            
    return resumen

#prueba
analizar_ventas(ventas)
```

```python
#Ejercicio 2
def validar_password(password):
    errores = []
    
    if len(password) < 8:
        errores.append("Debe tener al menos 8 caracteres.")
        
    tiene_mayus = False
    tiene_minus = False
    tiene_num = False
    tiene_simbolo = False
    tiene_espacio = False
    simbolos = "@#$%"
    
    for char in password:
        if char >= 'A' and char <= 'Z':
            tiene_mayus = True
        if char >= 'a' and char <= 'z':
            tiene_minus = True
        if char >= '0' and char <= '9':
            tiene_num = True
        if char in simbolos:
            tiene_simbolo = True
        if char == " ":
            tiene_espacio = True
            
    if not tiene_mayus:
        errores.append("Debe contener al menos una letra mayúscula.")
    if not tiene_minus:
        errores.append("Debe contener al menos una letra minúscula.")
    if not tiene_num:
        errores.append("Debe contener al menos un número.")
    if not tiene_simbolo:
        errores.append("Debe incluir al menos uno de los símbolos: @, #, $, %.")
    if tiene_espacio:
        errores.append("No debe contener espacios.")
        
    es_valida = (len(errores) == 0)
    return (es_valida, errores)

#Prueba
es_valida, errores = validar_password('Python123') 
print(es_valida) 
print(errores)
```

```python
#Ejercicio 3
def registrar_estudiante(estudiantes, nombre, notas):
    if nombre == "" or nombre in estudiantes:
        return "Error: El nombre es inválido o ya se encuentra registrado."
    
    if len(notas) < 3:
        return "Error: El estudiante debe registrar al menos tres notas."
    
    suma_notas = 0
    for nota in notas:
        if type(nota) != int and type(nota) != float:
            return "Error: Todas las notas deben ser numéricas y estar entre 0 y 20."
        if nota < 0 or nota > 20:
            return "Error: Todas las notas deben ser numéricas y estar entre 0 y 20."
        suma_notas += nota
        
    promedio_exacto = suma_notas / len(notas)
    promedio_dos_decimales = int(promedio_exacto * 100) / 100
    
    estado = "Aprobado" if promedio_dos_decimales >= 11 else "Desaprobado"
    
    estudiantes[nombre] = {
        "notas": notas,
        "promedio": promedio_dos_decimales,
        "estado": estado
    }
    
    return f"Registro exitoso del estudiante {nombre}."
#Prueba
estudiantes = {} 
registrar_estudiante( estudiantes, 'Ana', [18, 15, 17] )
```

```python
#Ejercicio 4
def agrupar_por_longitud(palabras):
    agrupados = {}
    vistos = []
    
    for p in palabras:
        if type(p) != str:
            continue
            
        p_minusc = p.lower()
        if p_minusc in vistos:
            continue
        vistos.append(p_minusc)
        
        longitud = len(p)
        if longitud not in agrupados:
            agrupados[longitud] = []
        agrupados[longitud].append(p)
        
    for longitud in agrupados:
        agrupados[longitud].sort()
        
    return agrupados
#Prueba
palabras = [ 'python', 'Java', 'código', 'web', 'java', 'datos', 'API', 'lista' ] 
resultado = agrupar_por_longitud(palabras)
print(resultado)
```

```python
#Ejercicio 5
def actualizar_inventario(inventario, producto, cantidad, operacion):
    if operacion != 'entrada' and operacion != 'salida':
        return {"mensaje": "Error: Tipo de operación no permitida."}
        
    if type(cantidad) != int or cantidad <= 0:
        return {"mensaje": "Error: La cantidad debe ser un entero positivo."}
        
    if operacion == 'entrada':
        if producto in inventario:
            inventario[producto] += cantidad
        else:
            inventario[producto] = cantidad
        return {"mensaje": "Entrada procesada.", "stock_actual": inventario[producto]}
        
    else:
        if producto not in inventario:
            return {"mensaje": "Error: Producto no registrado."}
        if inventario[producto] < cantidad:
            return {"mensaje": "Error: Stock insuficiente."}
            
        inventario[producto] -= cantidad
        return {"mensaje": "Salida procesada.", "stock_actual": inventario[producto]}
    #prueba
inventario = { 'Laptop': 5, 'Mouse': 12, 'Teclado': 8 } 
actualizar_inventario( inventario, 'Mouse', 4, 'salida' )
```

```python
#Ejercicio 6
def calcular_pedido(productos, cantidades, descuento=0, delivery=0):
    if descuento < 0 or descuento > 100:
        return {"error": "El descuento debe ser un porcentaje entre 0 y 100."}
    if delivery < 0:
        return {"error": "El costo de delivery debe ser mayor o igual a cero."}
        
    subtotal = 0
    no_encontrados = []
    
    for prod, cant in cantidades.items():
        if type(cant) != int or cant <= 0:
            return {"error": f"La cantidad asignada a '{prod}' debe ser un entero positivo."}
            
        if prod in productos:
            subtotal += productos[prod] * cant
        else:
            no_encontrados.append(prod)
            
    monto_descuento = subtotal * (descuento / 100)
    total_final = (subtotal - monto_descuento) + delivery
    
    return {
        "subtotal": subtotal,
        "descuento": monto_descuento,
        "delivery": delivery,
        "total": total_final,
        "no_encontrados": no_encontrados
    }
#Prueba
productos = {'Hamburguesa': 15, 'Pizza': 25, 'Gaseosa': 5}
cantidades = {'Hamburguesa': 2, 'Gaseosa': 3, 'Helado': 1}
calcular_pedido(productos, cantidades)
```

```python
#Ejercicio 7
def evaluar_empleados(empleados, minimo=70):
    resultado = []
    
    for nombre, puntajes in empleados.items():
        if len(puntajes) == 0:
            continue
            
        suma = 0
        for p in puntajes:
            suma += p
        promedio = suma / len(puntajes)
        
        if promedio >= 90:
            clasificacion = "Excelente"
        elif promedio >= 80:
            clasificacion = "Bueno"
        elif promedio >= 70:
            clasificacion = "Regular"
        else:
            clasificacion = "Deficiente"
            
        if promedio >= minimo:
            resultado.append((nombre, promedio, clasificacion))
            
    resultado.sort(key=lambda emp: (-emp[1], emp[0]))
    return resultado

#Prueba
empleados = { 'Ana': [80, 90, 75], 'Carlos': [60, 70, 65], 'Lucía': [95, 85, 90], 'Pedro': [70, 70, 70] }
evaluar_empleados(empleados)
```

```python
#Ejercicio 8
def comprimir_texto(texto):
    if len(texto) == 0:
        return ""
        
    resultado = ""
    letra_actual = texto[0]
    conteo = 1
    
    for i in range(1, len(texto)):
        letra = texto[i]
        if letra == letra_actual:
            conteo += 1
        else:
            resultado += letra_actual + str(conteo)
            letra_actual = letra
            conteo = 1
    resultado += letra_actual + str(conteo)
    
    return resultado


def descomprimir_texto(texto):
    if len(texto) == 0:
        return ""
        
    resultado = ""
    i = 0
    n = len(texto)
    
    while i < n:
        letra = texto[i]
        if not (letra >= 'a' and letra <= 'z') and not (letra >= 'A' and letra <= 'Z'):
            return "Error: Formato de compresion invalido."
            
        i += 1
        num_str = ""
        
        while i < n and (texto[i] >= '0' and texto[i] <= '9'):
            num_str += texto[i]
            i += 1
            
        if num_str == "":
            return "Error: Formato de compresion invalido."
            
        resultado += letra * int(num_str)
        
    return resultado

#Prueba
texto = 'aaabbccccdaa' 
comprimido = comprimir_texto(texto) 
print(comprimido)
original = descomprimir_texto(comprimido)
print(original)
```

```python
#Ejercicio 9

def transferir(cuentas, origen, destino, monto):
    if origen not in cuentas or destino not in cuentas:
        return {"estado": "Error", "mensaje": "Alguna de las cuentas ingresadas no existe."}
        
    if origen == destino:
        return {"estado": "Error", "mensaje": "Las cuentas de origen y destino no pueden ser iguales."}
        
    if (type(monto) != int and type(monto) != float) or monto <= 0:
        return {"estado": "Error", "mensaje": "El monto a transferir debe ser numérico y mayor a cero."}
        
    if cuentas[origen]["saldo"] < monto:
        return {"estado": "Error", "mensaje": "La cuenta de origen no posee saldo suficiente."}
        
    cuentas[origen]["saldo"] -= monto
    cuentas[destino]["saldo"] += monto
    
    return {
        "estado": "Exitoso",
        "titular_origen": cuentas[origen]["titular"],
        "titular_destino": cuentas[destino]["titular"],
        "monto_transferido": monto,
        "saldo_origen": cuentas[origen]["saldo"]
    }
#Prueba
cuentas = { '001': {'titular': 'Ana', 'saldo': 1500}, '002': {'titular': 'Luis', 'saldo': 800} }
transferir(cuentas, '001', '002', 300)
```

```python
#Ejercicio 10
def reservar_asientos(sala, asientos_solicitados):
    filas = len(sala)
    columnas = len(sala[0]) if filas > 0 else 0
    
    for f, c in asientos_solicitados:
        f_idx = f - 1
        c_idx = c - 1
        
        if f_idx < 0 or f_idx >= filas or c_idx < 0 or c_idx >= columnas:
            return {"estado": "Error", "mensaje": f"El asiento ({f}, {c}) no existe."}
            
        if sala[f_idx][c_idx] != 'L':
            return {"estado": "Error", "mensaje": f"El asiento ({f}, {c}) esta ocupado."}
            
    for f, c in asientos_solicitados:
        sala[f - 1][c - 1] = 'O'
        
    return {
        "estado": "Reserva Exitosa",
        "asientos_reservados": asientos_solicitados,
        "sala": sala
    }

def mostrar_sala(sala):
    for idx, fila in enumerate(sala):
        fila_texto = f"Fila {idx + 1}: "
        for asiento in fila:
            fila_texto += asiento + " "
        print(fila_texto)
#Prueba
sala = [['L', 'L', 'O', 'L'], ['O', 'L', 'L', 'L'], ['L', 'O', 'L', 'L']] 
asientos = [(1, 2), (2, 3), (3, 1)]
print(reservar_asientos(sala, asientos))
sala = [['L', 'L', 'O', 'L'], ['O', 'L', 'L', 'L'], ['L', 'O', 'L', 'L']] 
mostrar_sala(sala)
```
