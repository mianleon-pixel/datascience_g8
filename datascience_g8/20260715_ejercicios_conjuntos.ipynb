```python
#Ejercicio 1
estudiantes_python = {"Ana", "Luis", "María", "Carlos", "Sofía", "Diego"}
estudiantes_javascript = {"María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria"}

ambos = estudiantes_python.intersection(estudiantes_javascript)
solo_python = estudiantes_python.difference(estudiantes_javascript)
solo_javascript = estudiantes_javascript.difference(estudiantes_python)
todos = estudiantes_python.union(estudiantes_javascript)

print("Estudiantes en ambos cursos:", ambos)
print("Solo en Python:", solo_python)
print("Solo en JavaScript:", solo_javascript)
print("Todos los estudiantes:", todos)
```

```python
#Ejercicio 2
almacen_arequipa = {"laptop", "teclado", "mouse", "monitor", "impresora", "audifonos"}
almacen_lima = {"laptop", "mouse", "monitor", "tablet", "webcam", "parlantes"}

disponibles_ambos = almacen_arequipa.intersection(almacen_lima)
exclusivos_arequipa = almacen_arequipa.difference(almacen_lima)
exclusivos_lima = almacen_lima.difference(almacen_arequipa)
todos_productos = almacen_arequipa.union(almacen_lima)
mismo_inventario = almacen_arequipa == almacen_lima

print("Disponibles en ambos almacenes:", disponibles_ambos)
print("Exclusivos de Arequipa:", exclusivos_arequipa)
print("Exclusivos de Lima:", exclusivos_lima)
print("Todos los productos:", todos_productos)
print("¿Tienen el mismo inventario?:", mismo_inventario)
```

```python
#Ejercicio 3
permisos_usuario = {"leer", "editar", "comentar", "descargar"}
permisos_requeridos = {"leer", "editar", "eliminar", "publicar"}

posee_todos = permisos_requeridos.issubset(permisos_usuario)
permisos_faltantes = permisos_requeridos.difference(permisos_usuario)
permisos_adicionales = permisos_usuario.difference(permisos_requeridos)
total_diferentes = len(permisos_usuario.union(permisos_requeridos))

print("¿Posee todos los permisos requeridos?:", posee_todos)
print("Permisos que le faltan:", permisos_faltantes)
print("Permisos adicionales:", permisos_adicionales)
print("Cantidad total de permisos diferentes:", total_diferentes)
```

```python
#Ejercicio 4
numeros_a = {2, 4, 6, 8, 10, 12, 14}
numeros_b = {4, 6, 9, 10, 13, 14, 16}
numeros_c = {1, 4, 6, 10, 11, 14, 18}

en_tres = numeros_a.intersection(numeros_b, numeros_c)
en_a_y_b_no_c = numeros_a.intersection(numeros_b).difference(numeros_c)
solo_en_a = numeros_a.difference(numeros_b.union(numeros_c))
todos = numeros_a.union(numeros_b, numeros_c)

print("Presentes en los tres conjuntos:", en_tres)
print("En A y B, pero no en C:", en_a_y_b_no_c)
print("Solo pertenecen a A:", solo_en_a)
print("Todos los números diferentes:", todos)
```
