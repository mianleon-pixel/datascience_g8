```python
#Ejercicio 1
numeros = [4, 7, 2, 4, 9, 7, 1, 2, 10, 4]
# 2. Creamos una lista vacía para guardar los números únicos
unicos = []
for n in numeros:
    if n not in unicos:
        unicos.append(n)
print("Lista sin duplicados:", unicos)
```

```python
#Ejercicio 2
numeros = [12, 7, 9, 24, 31, 18, 5, 42, 13, 8]
pares = []
impares = []
for n in numeros:  
    if n % 2 == 0:
        pares.append(n)  
    else:
        impares.append(n)
print("Números pares:", pares)
print("Números impares:", impares)
```

```python
#Ejercicio 3
numeros = [45, 12, 89, 34, 89, 23, 67, 10]
numeros_unicos = list(set(numeros))
numeros_ordenados = sorted(numeros_unicos)
segundo_mayor = numeros_ordenados[-2]
print("Lista ordenada sin repetidos:", numeros_ordenados)
print("El segundo número mayor es:", segundo_mayor)
```

```python
#Ejercicio 4
palabras = ['python', 'sol', 'programacion', 'lista', 'computadora', 'if', 'variable']
contador_largas = 0
for palabra in palabras:
       if len(palabra) > 5:
              contador_largas = contador_largas + 1 
print("La cantidad de palabras con más de 5 letras es:", contador_largas)
```

```python
#Ejercicio 5
notas = [14, 18, 11, 16, 9, 13, 20]
suma_notas = sum(notas)
cantidad_notas = len(notas)
promedio = suma_notas / cantidad_notas
print("El promedio del estudiante es:", round(promedio, 2))
if promedio >= 11:
    print("Estado: APROBADO")
else:
    print("Estado: DESAPROBADO")
```

```python
#Ejercicio 7
numeros = [3, 6, 9, 12, 15]
duplicados = []
for n in numeros:
       resultado = n * 2
       duplicados.append(resultado)
print("Lista original:", numeros)
print("Lista multiplicada por 2:", duplicados)
```

```python
#Ejercicio 8
lista1 = [1, 3, 5, 7, 9, 11]
lista2 = [2, 3, 6, 7, 10, 11]
comunes = []
for n in lista1:   
    if n in lista2:        
        comunes.append(n)
print("Elementos comunes en ambas listas:", comunes)
```

```python
#Ejercicio 9
lista1 = [4, 8, 12, 16, 20]
lista2 = [8, 10, 12, 18, 20]
unicos = []
for n in lista1:
    if n not in lista2:
        unicos.append(n)
for n in lista2:
    if n not in lista1:
        unicos.append(n)
print("Elementos únicos entre ambas listas:", unicos)
```

```python
#Ejercicio 13
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'Maria', 'Carlos']
nombres_unicos = set(nombres)
for nombre in nombres_unicos:   
    apariciones = nombres.count(nombre)
    print(f"El nombre {nombre} aparece {apariciones} veces.")
```

```python
#Ejercicio 15
frutas = ['manzana', 'pera', 'uva', 'sandía', 'plátano', 'mango']
busqueda = input("Escribe el nombre de una fruta: ").lower().strip()
if busqueda in frutas:
    posicion = frutas.index(busqueda)
    print(f"La fruta '{busqueda}' está en la posición {posicion}.")
else:
    print(f"La fruta '{busqueda}' no se encuentra en la lista.")
```

```python
#Ejercicio 19
numeros = [1, 2, 3, 4]
letras = ['a', 'b', 'c', 'd']
combinada = []

for i in range(len(numeros)):
    combinada.append(numeros[i])
    combinada.append(letras[i])

print("Lista combinada alternada:", combinada)
```

```python
#Ejercicio 20
estudiantes = ['Ana', 'Luis', 'Carlos', 'María', 'Pedro']
notas = [18, 10, 14, 8, 16]
aprobados = []

for i in range(len(estudiantes)):
    if notas[i] >= 11:
        aprobados.append(estudiantes[i])

print("Estudiantes aprobados:", aprobados)
```
