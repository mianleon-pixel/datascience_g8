```python
#Ejercicio 1
numeros = (15, 8, 42, 3, 27, 19, 5)

mayor = max(numeros)
menor = min(numeros)
diferencia = mayor - menor

print(f"Número mayor: {mayor}")
print(f"Número menor: {menor}")
print(f"Diferencia: {diferencia}")
```

```python
#Ejercicio 2
datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3)

numero = int(input("Ingresa un número para buscar: "))
cantidad = datos.count(numero)

print(f"El número {numero} aparece {cantidad} veces.")
```

```python
#Ejercicio 3
nombres = ('Ana', 'Luis', 'Pedro', 'María', 'Luis', 'Carlos')

busqueda = input("Ingresa el nombre a buscar: ")

if busqueda in nombres:
    posicion = nombres.index(busqueda)
    print(f"{busqueda} se encuentra en el índice {posicion}.")
else:
    print(f"El nombre '{busqueda}' no existe en la tupla.")
```

```python
#Ejercicio 4
notas = (18, 15, 20, 11, 14, 17, 19)

promedio = sum(notas) / len(notas)
maxima = max(notas)
minima = min(notas)
aprobados = len([n for n in notas if n >= 11])

print(f"Promedio: {promedio:.2f}")
print(f"Nota máxima: {maxima}")
print(f"Nota mínima: {minima}")
print(f"Cantidad de aprobados: {aprobados}")
```

```python
#Ejercicio 5
ventas = [('Lunes', 120), ('Martes', 150), ('Miércoles', 95), ('Jueves', 180), ('Viernes', 210)]

# Asumimos inicialmente que el primer día es el mayor
dia_max = ventas[0][0]
venta_max = ventas[0][1]
total = 0

# Desempaquetamos directamente 'dia' y 'venta' en el bucle
for dia, venta in ventas:
    total += venta
    if venta > venta_max:
        venta_max = venta
        dia_max = dia

promedio = total / len(ventas)

print(f"Día con mayor venta: {dia_max} con {venta_max}")
print(f"Total de ventas: {total}")
print(f"Promedio de ventas: {promedio:.2f}")
```
